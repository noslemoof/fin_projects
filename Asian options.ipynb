### Asian Call, binomial tree Iteration

```python
import numpy as np

'''params'''
t = 1
n = 20
dt = t/n
S_0 = 100
sigma = 0.2
r=0.05
x = 100

# populate averages in terminal nested list [average,terminal stock,net up moves]
def average_prices(S_0,n,t,sigma):
    dt = t/n
    u = np.exp(np.sqrt(dt)*sigma)
    d = 1/u

    step_i = [(S_0,S_0,0)]
    for i in range(1,n+1,1):
        temp = []
        for j in step_i:
            temp.append([j[1]*u/(i+1)+j[0]*i/(i+1),j[1]*u,j[2]+1])
            temp.append([j[1]*d/(i+1)+j[0]*i/(i+1),j[1]*d,j[2]-1])
        step_i= temp

    return step_i
```

### Floating Strike

```python
def FloatingStrike_asian_call(S_0,n,t,sigma,r):
    dt = t/n
    u = np.exp(np.sqrt(dt)*sigma)
    d = 1/u
    p = (np.exp(r*dt)-d)/(u-d)

    average_list = average_prices(S_0,n,t,sigma)

    final = 0
    for i in average_list:
        final+= np.maximum(0,i[1]-i[0]) * p**((n+i[2])/2) * (1-p)**((n-i[2])/2) * np.exp(-r*t)

    return final
    

FloatingStrike_asian_call(S_0,n,t,sigma,r)
```

### Fixed Strike

```python
def FixedStrike_asian_call(S_0,x,n,t,sigma,r):
    dt = t/n
    u = np.exp(np.sqrt(dt)*sigma)
    d = 1/u
    p = (np.exp(r*dt)-d)/(u-d)

    average_list = average_prices(S_0,n,t,sigma)

    final = 0
    for i in average_list:
        final+= np.maximum(0,i[0]-x) * p**((n+i[2])/2) * (1-p)**((n-i[2])/2) * np.exp(-r*t)

    return final
    
FixedStrike_asian_call(S_0,x,n,t,sigma,r)
```
